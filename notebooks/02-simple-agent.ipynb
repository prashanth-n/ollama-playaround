```python
from dotenv import load_dotenv
from ollama import Client
import os
load_dotenv()

OLLAMA_API_KEY = os.environ.get("OLLAMA_CLOUD_API_KEY")
client = Client(
    host="https://ollama.com",
    headers={'Authorization': 'Bearer ' + OLLAMA_API_KEY}
)

def get_value_from_percentage(base, percentage):
    return base * percentage / 100

AVAILABLE_TOOLS = {
    "calculate_percentage": get_value_from_percentage
}

TOOLS_SCHEMA =[
    {
        "type": "function",
        "function": {
            "name": "calculate_percentage",
            "description": "Calculate percentage of a base number",
            "parameters": {
                "type": "object",
                "properties": {
                    "base": {"type": "number"},
                    "percent": {"type": "number"}
                },
                "required": ["base", "percent"]
            }
        }
    }
]

response = client.chat(
        model="glm-5",
        messages=[{"role": "user", "content": f""" Compare 10% of 5000 and 34% of 56778483 """}],
        format="json",
        options={"temperature": 0},
        tools=TOOLS_SCHEMA
    )
print(response.message)
```
